# Sample scRNA-seq analysis (M3 parser fixture)
Synthetic notebook for signature-driven extraction tests.

```python
import scanpy as sc
adata = sc.read_h5ad("data/GSE131907.h5ad")
sc.pp.filter_cells(adata, min_genes=200, min_counts=500)
```

```python
sc.pp.normalize_total(adata, target_sum=1e4)
sc.pp.log1p(adata)
```

```python
sc.pp.highly_variable_genes(adata, flavor="seurat_v3", n_top_genes=2000)
```

```python
sc.pp.pca(adata, n_comps=50)
sc.pp.neighbors(adata)
```

```python
sc.tl.leiden(adata, resolution=0.8)
```

```python
sc.tl.umap(adata)
```

```python
sc.tl.rank_genes_groups(adata, groupby="cell_type", method="wilcoxon")
```
